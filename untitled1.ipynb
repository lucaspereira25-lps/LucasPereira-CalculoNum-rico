### Problema

Fluido newtoniano entre duas placas paralelas separadas por $H = 0{,}01$ m. Em $t = 0$ a placa em $y = 0$ passa a se mover com $U = 1$ m/s e a placa em $y = H$ permanece fixa. O escoamento é unidirecional, sem gradiente de pressão, governado pela equação da difusão:

$$\frac{\partial u}{\partial t} = \nu \frac{\partial^2 u}{\partial y^2}$$

com $u(0,t) = U$, $u(H,t) = 0$ e $u(y,0) = 0$ no interior.

### Discretização (Euler explícito no tempo + diferenças centrais no espaço)

$$u_i^{n+1} = u_i^n + r\left(u_{i+1}^n - 2u_i^n + u_{i-1}^n\right), \qquad r = \frac{\nu\,\Delta t}{\Delta y^2}$$

Reorganizando:

$$u_i^{n+1} = r\,u_{i-1}^n + (1-2r)\,u_i^n + r\,u_{i+1}^n$$

### Condição de estabilidade

$$r \le \frac{1}{2} \qquad \Longleftrightarrow \qquad \Delta t \le \frac{\Delta y^2}{2\nu}$$

## 1. Verificação da estabilidade (cálculo analítico, antes de executar)

Dados: $H = 0{,}01$ m, $\nu = 10^{-4}$ m²/s, 21 pontos incluindo as paredes.

$$\Delta y = \frac{H}{21-1} = \frac{0{,}01}{20} = 5\times10^{-4}\ \text{m}$$

$$\Delta t_{max} = \frac{\Delta y^2}{2\nu} = \frac{(5\times10^{-4})^2}{2\times10^{-4}} = \frac{2{,}5\times10^{-7}}{2\times10^{-4}} = 1{,}25\times10^{-3}\ \text{s}$$

| Caso | $\Delta t$ (s) | $r = \nu\Delta t/\Delta y^2$ | $r \le 1/2$ ? | Classificação prevista |
|------|--------|-------------|-----------|------------------------|
| A | 0,001 | 0,40 | sim | **estável** |
| B | 0,002 | 0,80 | não | **instável** |

No caso B temos $1 - 2r = 1 - 1{,}6 = -0{,}6 < 0$: o coeficiente central fica negativo, a atualização deixa de ser uma média ponderada e o modo de maior frequência da malha (oscilação ponto a ponto, "serrilhado") é amplificado a cada passo pelo fator $|1 - 4r| = |1 - 3{,}2| = 2{,}2$.

Note que $\Delta t = 0{,}002$ s é $1{,}6\times$ maior que $\Delta t_{max} = 0{,}00125$ s.

```python
import numpy as np
import matplotlib.pyplot as plt


def simular(dt, tipo=np.float32, passos=2000, pontos=21):
    """Simula difusao de velocidade e detecta falhas."""
    H = 0.01
    nu = 1e-4
    y = np.linspace(0, H, pontos)
    dy = H / (pontos - 1)

    # Coeficientes no tipo numerico escolhido.
    r = tipo(nu * dt / dy**2)
    dois = tipo(2)

    # Fluido parado e placa em y=0 em movimento.
    u = np.zeros(pontos, dtype=tipo)
    u[0] = tipo(1)

    tempos = [0.0]
    maximos = [float(np.max(np.abs(u)))]
    falha = None

    # Transforma avisos numericos em excecoes.
    with np.errstate(over="raise", invalid="raise"):
        for passo in range(1, passos + 1):
            try:
                novo = u.copy()
                # Atualiza o interior e preserva as paredes.
                novo[1:-1] = (
                    u[1:-1]
                    + r * (
                        u[2:] - dois * u[1:-1] + u[:-2]
                    )
                )
            except FloatingPointError as erro:
                falha = passo
                print(
                    f"dt={dt}, {tipo.__name__}: "
                    f"falha no passo {passo}, "
                    f"t={passo * dt:.4f} s - {erro}"
                )
                break

            u = novo
            tempos.append(passo * dt)
            maximos.append(float(np.max(np.abs(u))))

    if falha is None:
        print(
            f"dt={dt}, {tipo.__name__}: "
            f"{passos} passos sem overflow."
        )

    return {
        "y": y,
        "u": u,
        "tempos": tempos,
        "maximos": maximos,
        "r": float(r),
        "falha": falha,
    }
```

## 2. Detecção de overflow

Execução dos quatro testes: os dois passos de tempo combinados com `float32` e `float64`.

```python
print("Maior float32:", np.finfo(np.float32).max)
print("Maior float64:", np.finfo(np.float64).max)

resultados = {}
for tipo in (np.float32, np.float64):
    for dt in (0.001, 0.002):
        nome = f"{tipo.__name__}, dt={dt}"
        resultados[nome] = simular(dt, tipo)

plt.figure(figsize=(9, 5))
for nome, resultado in resultados.items():
    plt.semilogy(
        resultado["tempos"],
        resultado["maximos"],
        label=nome
    )
plt.axhline(1.0, color="k", ls="--", lw=1,
            label="limite fisico |u| = U = 1 m/s")
plt.xlabel("Tempo simulado (s)")
plt.ylabel("Maximo de |u| (m/s) - escala logaritmica")
plt.title("Crescimento da velocidade e overflow")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import pandas as pd

linhas = []
for nome, res in resultados.items():
    tipo_txt, dt_txt = nome.split(", dt=")
    linhas.append({
        "tipo numerico": tipo_txt,
        "dt (s)": float(dt_txt),
        "r": round(res["r"], 4),
        "r <= 0.5": "sim" if res["r"] <= 0.5 else "nao",
        "overflow": "nao" if res["falha"] is None else "sim",
        "iteracao da falha": "-" if res["falha"] is None else res["falha"],
        "t da falha (s)": "-" if res["falha"] is None
                          else round(res["falha"] * float(dt_txt), 4),
        "max|u| final": f"{res['maximos'][-1]:.3e}",
    })

tabela = pd.DataFrame(linhas)
display(tabela)
```

As previsões da Atividade 1 se confirmaram: os dois casos com $r = 0{,}4$ concluíram os 2000 passos com $\max|u| = 1$ m/s, e os dois casos com $r = 0{,}8$ falharam por overflow.

```python
# Evidencia de que a solucao ja esta errada MUITO antes do overflow.
instavel = resultados["float32, dt=0.002"]
m = instavel["maximos"]

print("Caso instavel (float32, dt = 0.002 s, r = 0.8):\n")
print(f"{'max|u| ultrapassa':>20} | {'passo':>6} | {'t (s)':>7}")
print("-" * 40)
for lim in (1.1, 10, 1e3, 1e10, 1e30):
    idx = next((i for i, v in enumerate(m) if v > lim), None)
    if idx is not None:
        print(f"{lim:>20g} | {idx:>6} | {idx * 0.002:>7.3f}")
print(f"{'overflow (3.4e38)':>20} | {instavel['falha']:>6} | "
      f"{instavel['falha'] * 0.002:>7.3f}")
```

## 3. Interpretação física

Fisicamente, o único mecanismo de transporte aqui é a **difusão viscosa**, que espalha e suaviza quantidade de movimento a partir da placa. Como a placa móvel injeta no máximo $U = 1$ m/s e a placa fixa impõe $0$ m/s, o princípio do máximo garante que a solução exata satisfaz

$$0 \le u(y,t) \le U = 1\ \text{m/s} \qquad \text{para todo } y \text{ e todo } t.$$

O esquema explícito só reproduz essa propriedade quando $0 \le r \le 1/2$, pois aí

$$u_i^{n+1} = r\,u_{i-1}^n + (1-2r)\,u_i^n + r\,u_{i+1}^n$$

é uma **média ponderada** — coeficientes não negativos que somam 1 — e nenhum valor novo pode sair do intervalo definido pelos vizinhos.

Com $r = 0{,}8$ o coeficiente central $(1-2r) = -0{,}6$ é negativo. O modo serrilhado ($u_i \propto (-1)^i$) é multiplicado por $|1 - 4r| = 2{,}2$ a cada passo, ou seja, cresce **exponencialmente**. No gráfico monolog isso aparece como uma **reta ascendente**, assinatura inconfundível de crescimento geométrico; no caso estável a curva satura em 1 e fica horizontal.

**O ponto central da atividade:** no caso instável, $\max|u|$ já ultrapassa 1 m/s no **passo 3** ($t = 0{,}006$ s) e passa de 10 m/s no **passo 8** ($t = 0{,}016$ s) — muito antes do overflow no passo 121. Ou seja, a solução já era fisicamente absurda quase desde o início, e o overflow é apenas o sintoma final e tardio.

> **Instabilidade numérica $\neq$ overflow.** A instabilidade amplifica perturbações desde o primeiro passo; o overflow só acontece quando o crescimento esbarra no limite de representação do tipo numérico.

## 4. Capacidade de representação (float32 × float64)

**Não — trocar o tipo numérico não elimina a instabilidade; apenas adia o overflow.**

O valor de $r$ depende somente de $\nu$, $\Delta t$ e $\Delta y$: ele não muda com a precisão. A taxa de crescimento por passo continua sendo $2{,}2$ nos dois tipos. O que muda é **até onde** o número cabe antes de estourar:

| tipo | maior valor finito | passo da falha |
|---|---|---|
| float32 | $3{,}4\times10^{38}$ | 121 |
| float64 | $1{,}8\times10^{308}$ | 917 |

A previsão teórica confirma o mecanismo. Partindo de amplitude $\sim 1$ e crescendo por $2{,}2$ a cada passo, o número de passos até o limite é $n \approx \ln(\text{máximo})/\ln(2{,}2)$:

- **float32:** $\ln(3{,}4\times10^{38})/\ln(2{,}2) = 88{,}7/0{,}789 \approx 113$ passos (observado: **121**)
- **float64:** $\ln(1{,}8\times10^{308})/\ln(2{,}2) = 709{,}2/0{,}789 \approx 900$ passos (observado: **917**)

A razão $917/121 \approx 7{,}6$ é aproximadamente a razão entre os expoentes máximos ($308/38{,}5 \approx 8$) — e não um ganho de estabilidade. Em ambos os casos a solução já está errada nos primeiros passos: mais bits de expoente compram tempo, não correção. A cura é reduzir $r$, não trocar o tipo.

```python
# Verificacao numerica do argumento da Atividade 4.
fator = abs(1 - 4 * 0.8)   # amplificacao do modo serrilhado por passo
print(f"Fator de amplificacao por passo: |1 - 4r| = {fator}")

for nome_tipo, maximo in (("float32", np.finfo(np.float32).max),
                          ("float64", np.finfo(np.float64).max)):
    n = np.log(maximo) / np.log(fator)
    print(f"  {nome_tipo}: passos previstos ate o overflow = {n:.0f}")

print("\nObservado: float32 -> passo 121 | float64 -> passo 917")
```

## 5. Refinamento da malha (41 pontos)

Com 41 pontos:

$$\Delta y = \frac{H}{41-1} = \frac{0{,}01}{40} = 2{,}5\times10^{-4}\ \text{m} \quad \text{(metade do anterior)}$$

$$\Delta t_{max} = \frac{\Delta y^2}{2\nu} = \frac{(2{,}5\times10^{-4})^2}{2\times10^{-4}} = 3{,}125\times10^{-4}\ \text{s} \quad \text{(um quarto do anterior)}$$

Mantendo $\Delta t = 0{,}001$ s — que era **estável** na malha de 21 pontos — agora $r = 1{,}6 > 0{,}5$, portanto **instável**.

```python
# --- Atividade 5: refinamento da malha (41 pontos) ---
H, nu = 0.01, 1e-4
pontos = 41
dy41 = H / (pontos - 1)
dt_max41 = dy41**2 / (2 * nu)

print(f"dy    = {dy41:.6f} m")
print(f"dtmax = {dt_max41:.3e} s")
for dt in (0.001, 0.0002):
    print(f"   dt={dt}: r = {nu * dt / dy41**2:.3f}")

print("\n-> dt = 0.001 s (r = 1.6, instavel):")
ref_instavel = simular(0.001, np.float64, passos=2000, pontos=41)

print("\n-> dt = 0.0002 s (r = 0.32, estavel):")
ref_estavel = simular(0.0002, np.float64, passos=2000, pontos=41)

plt.figure(figsize=(9, 5))
plt.semilogy(ref_instavel["tempos"], ref_instavel["maximos"],
             label="41 pontos, dt=0.001 (r=1.6)")
plt.semilogy(ref_estavel["tempos"], ref_estavel["maximos"],
             label="41 pontos, dt=0.0002 (r=0.32)")
plt.axhline(1.0, color="k", ls="--", lw=1, label="limite fisico U = 1 m/s")
plt.xlabel("Tempo simulado (s)")
plt.ylabel("Maximo de |u| (m/s) - escala logaritmica")
plt.title("Efeito do refinamento da malha")
plt.grid(True)
plt.legend()
plt.show()
```

### Resposta da Atividade 5

Resultados obtidos:

| pontos | $\Delta y$ (m) | $\Delta t$ (s) | $r$ | resultado |
|---|---|---|---|---|
| 41 | 2,5e-4 | 0,001 | 1,60 | **overflow no passo 426** ($t = 0{,}426$ s) |
| 41 | 2,5e-4 | 0,0002 | 0,32 | 2000 passos sem overflow, $\max\|u\| = 1$ |

O ponto essencial é que **refinar a malha piora a restrição de estabilidade**. Como $\Delta t_{max} \propto \Delta y^2$, dividir $\Delta y$ por 2 exige dividir $\Delta t$ por 4. Refinar o espaço "de graça", sem mexer no tempo, quebra a simulação — foi exatamente o que aconteceu com $\Delta t = 0{,}001$ s, um passo que funcionava perfeitamente na malha grossa.

Escolhendo $\Delta t = 2\times10^{-4}\ \text{s} < 3{,}125\times10^{-4}$ s temos $r = 0{,}32 \le 0{,}5$: a simulação completa 2000 passos sem overflow, $\max|u|$ permanece em 1 m/s e a solução é limitada e fisicamente coerente.

Observe o custo: 2000 passos agora correspondem a apenas $t = 0{,}4$ s de tempo físico, contra 2 s antes. Mais resolução espacial exige **muito mais passos** para simular o mesmo intervalo de tempo — é o preço do método explícito.

## 6. Verificação do perfil de velocidade

A solução estacionária (perfil de Couette) é linear:

$$u_{est}(y) = U\left(1 - \frac{y}{H}\right)$$

e o erro máximo é medido por

$$E_\infty = \max_i \left|u_i - u_{est}(y_i)\right|$$

A escala de tempo difusiva do problema é $\tau = H^2/\nu = (0{,}01)^2/10^{-4} = 1$ s, então são necessários vários múltiplos de $\tau$ para atingir o regime estacionário.

```python
# --- Atividade 6: perfil final x solucao estacionaria ---
U, H = 1.0, 0.01

casos = {
    "t = 0,2 s (200 passos)":  simular(0.001, np.float64, passos=200),
    "t = 1,0 s (1000 passos)": simular(0.001, np.float64, passos=1000),
    "t = 3,0 s (3000 passos)": simular(0.001, np.float64, passos=3000),
}

y = casos["t = 3,0 s (3000 passos)"]["y"]
u_est = U * (1 - y / H)

plt.figure(figsize=(7, 6))
for nome, res in casos.items():
    plt.plot(res["u"], res["y"], "o-", ms=4, label=nome)
plt.plot(u_est, y, "k--", lw=2, label="estacionario: U(1 - y/H)")
plt.xlabel("u (m/s)")
plt.ylabel("y (m)")
plt.title("Evolucao do perfil de velocidade ate o regime estacionario")
plt.grid(True)
plt.legend()
plt.show()

print("Erro maximo em relacao ao perfil estacionario:")
for nome, res in casos.items():
    Einf = np.max(np.abs(res["u"] - u_est))
    print(f"   {nome:26s} -> E_inf = {Einf:.4e} m/s")
```

```python
# Efeito do tipo numerico sobre a PRECISAO (nao sobre a estabilidade).
e32 = np.max(np.abs(simular(0.001, np.float32, passos=2000)["u"] - u_est))
e64 = np.max(np.abs(simular(0.001, np.float64, passos=2000)["u"] - u_est))

print(f"\nEm t = 2 s (caso estavel, r = 0.4):")
print(f"   E_inf(float32) = {e32:.3e} m/s")
print(f"   E_inf(float64) = {e64:.3e} m/s")
print(f"   eps de maquina: float32 = {np.finfo(np.float32).eps:.2e}, "
      f"float64 = {np.finfo(np.float64).eps:.2e}")
```

### Resposta da Atividade 6

| Tempo simulado | Passos | $E_\infty$ (m/s) |
|---|---|---|
| 0,2 s | 200 | 8,78e-02 |
| 1,0 s | 1000 | 3,19e-05 |
| 3,0 s | 3000 | **8,08e-14** |

**O tempo simulado foi suficiente?** Depende do instante analisado. Em $t = 0{,}2$ s (apenas $0{,}2\,\tau$) o perfil ainda é fortemente **curvo**: a informação da placa móvel ainda não atravessou todo o canal, e $E_\infty \approx 0{,}088$ m/s — quase 9% de $U$ — é dominado pelo **transiente ainda não concluído**, não por erro numérico. Seria um equívoco interpretar esse valor como falta de precisão do método.

Já em $t = 3$ s (três escalas difusivas) o perfil coincide visualmente com a reta e $E_\infty \approx 8\times10^{-14}$, ou seja, no nível do arredondamento da dupla precisão. O regime estacionário foi efetivamente atingido.

Isso mostra que $E_\infty$ mistura **duas fontes**: (i) o transiente físico residual, que decai exponencialmente com o tempo, e (ii) o erro de discretização e arredondamento. Aqui, como o perfil estacionário é uma reta e a diferença central de segunda ordem é **exata** para funções lineares, o erro de truncamento espacial se anula no limite estacionário e sobra apenas o arredondamento. Por isso, em $t = 2$ s, o float64 chega a $1{,}6\times10^{-9}$ enquanto o float32 estaciona em $\approx 2{,}1\times10^{-6}$, compatível com seu épsilon de máquina ($1{,}2\times10^{-7}$) acumulado ao longo dos passos.

Conclusão parcial: **o tipo numérico afeta a precisão do resultado, não a estabilidade do método.**

## 7. Conclusão

O método de Euler explícito para a equação da difusão é **condicionalmente estável**: só funciona se

$$r = \frac{\nu\,\Delta t}{\Delta y^2} \le \frac{1}{2} \qquad \Longleftrightarrow \qquad \Delta t \le \frac{\Delta y^2}{2\nu}$$

Os três fenômenos investigados se encadeiam nesta ordem de causa e efeito:

**1. Malha → passo de tempo.** Refinar o espaço aperta o limite de forma **quadrática** ($\Delta t_{max} \propto \Delta y^2$). Passar de 21 para 41 pontos reduziu $\Delta y$ pela metade e $\Delta t_{max}$ a um quarto ($1{,}25\times10^{-3} \to 3{,}125\times10^{-4}$ s). Um $\Delta t$ que era seguro tornou-se instável sem nenhuma outra alteração no código.

**2. Passo de tempo → instabilidade.** Quando $r > 1/2$, o coeficiente central $(1-2r)$ fica negativo, a atualização deixa de ser uma média ponderada e o modo serrilhado da malha é multiplicado por $|1-4r|$ a cada passo. Erros de arredondamento — inevitáveis em ponto flutuante — viram sementes de um crescimento exponencial. A solução já estava errada no passo 3.

**3. Instabilidade → overflow.** O crescimento exponencial esbarra, cedo ou tarde, no maior valor representável do tipo ($3{,}4\times10^{38}$ no float32, $1{,}8\times10^{308}$ no float64). O overflow é a **consequência final**, não a causa. Aumentar a precisão adiou a falha de 121 para 917 passos, mas a solução era igualmente inútil nos dois casos.

**Lição prática:** o overflow é um sintoma barulhento e tardio de um problema silencioso e precoce. Uma simulação instável pode até terminar sem nenhum aviso e ainda assim produzir números inteiramente falsos. Por isso, monitorar limites físicos conhecidos ($0 \le u \le U$) é uma verificação mais confiável do que esperar por `inf` ou `NaN`.

**Qual alteração corrigiu a causa da falha?** Reduzir o passo de tempo até satisfazer $r \le 1/2$: $\Delta t = 0{,}001$ s com 21 pontos ($r = 0{,}4$) e $\Delta t = 2\times10^{-4}$ s com 41 pontos ($r = 0{,}32$). Alternativamente, poder-se-ia manter $\Delta t$ e engrossar a malha. O que **não** resolve é trocar `float32` por `float64`, pois a precisão não altera o valor de $r$. A causa da falha está no **método e em seu critério de estabilidade**, não na representação em ponto flutuante.
