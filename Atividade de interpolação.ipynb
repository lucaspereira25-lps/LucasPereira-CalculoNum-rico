## Seção 8
**Registro do grupo:**

- **Ponto alterado:** a medição em $x = 2$, que passou de $y = 2{,}7$ para $y = 3{,}2$ ($+0{,}5$).
- **Efeito em $x = 1{,}5$:** a estimativa foi de $2{,}3125$ para $2{,}59375$, um aumento de $0{,}28125$ (cerca de 12%), mesmo $x = 1{,}5$ não sendo o ponto alterado.
- **A curva mudou só perto de $x = 2$?** Não, mudou no intervalo **inteiro**, embora continue passando pelos outros três pontos. Entre 0 e 1 a curva nova ficou **abaixo** da original ($-0{,}156$ em $x = 0{,}5$); entre 2 e 3 ficou **acima** ($+0{,}469$ em $x = 2{,}5$). Os coeficientes também mudaram bastante: o de $x^3$ foi de $+0{,}133$ para $-0{,}117$ e trocou de sinal.
- **Explicação:** pela forma de Lagrange, a mudança na curva é $\Delta y \cdot L_2(x)$, em que $L_2$ é o polinômio de Lagrange do nó $x = 2$. Em $x = 1{,}5$, $L_2(1{,}5) = 0{,}5625$, e $0{,}5 \times 0{,}5625 = 0{,}28125$, exatamente a variação observada. A mudança é proporcional à alteração: $-0{,}5$ gera $-0{,}28125$ e $+0{,}1$ gera $+0{,}05625$ (testado na célula extra abaixo).
- **Por que pequenas incertezas importam:** como a curva é obrigada a passar exatamente por cada dado, um erro de medição em **um** ponto se propaga para as estimativas em **todo** o intervalo e muda os coeficientes. O interpolador não separa sinal de ruído. Por isso as incertezas das medições precisam ser conhecidas antes de confiar nas estimativas, principalmente com polinômios de grau alto.

## Célula extra

```python
# CÉLULA EXTRA – Outras alterações em y(2) e conferência pela forma de Lagrange
def L2(t):
    # polinômio de Lagrange associado ao nó x=2 (nós 0, 1, 2, 3)
    return (t - 0)*(t - 1)*(t - 3) / ((2 - 0)*(2 - 1)*(2 - 3))

print(f"L2(1.5) = {L2(1.5)}")
for delta in [0.5, -0.5, 0.1]:
    ya = yB.copy(); ya[2] += delta
    pa = np.polynomial.Polynomial.fit(xB, ya, deg=len(xB)-1).convert()
    print(f"Δy = {delta:+.1f} | P(1.5) = {pa(1.5):.5f} | variação = {pa(1.5)-pB(1.5):+.5f} "
          f"| Δy·L2(1.5) = {delta*L2(1.5):+.5f} | coef = {np.round(pa.coef, 4)}")
```

**Respostas do grupo:**

- **Todas passam pelos pontos?** Sim. Nos nós o erro é da ordem de $10^{-16}$ a $10^{-12}$ (arredondamento). O erro cresce um pouco com $n$ porque o problema fica mais mal condicionado.
- **Perto das extremidades:** no centro, a aproximação melhora com mais nós, mas perto de $x = \pm1$ as oscilações **aumentam** com $n$. O erro máximo em relação a $f$ foi de cerca de $0{,}44$ (5 nós), $1{,}05$ (9 nós) e $7{,}2$ (15 nós), ou seja, mais pontos deram um resultado **pior**. Esse é o fenômeno de Runge: com nós igualmente espaçados, o polinômio de grau alto oscila nas pontas.
- **Alternativas para muitos dados:**
  - **splines cúbicas** (interpolação por partes com polinômios de grau 3 e derivadas contínuas): com 15 nós, erro máximo $\approx 0{,}0025$;
  - **nós de Chebyshev** (mais concentrados nas extremidades): com 15 nós, erro máximo $\approx 0{,}047$;
  - interpolação linear por partes;
  - **regressão** com polinômio de grau baixo, se os dados tiverem ruído.

  *(Comparação na célula extra abaixo.)*

# Célula extra

```python
# CÉLULA EXTRA – Erro máximo e alternativas: nós de Chebyshev e spline cúbica
from scipy.interpolate import CubicSpline

print(f"{'n':>3} | {'erro nos nós':>12} | {'equiesp.':>9} | {'Chebyshev':>9} | {'spline':>9}")
for n in [5, 9, 15]:
    tn = np.linspace(-1, 1, n)
    pn = np.polynomial.Polynomial.fit(tn, f(tn), deg=n-1).convert()
    k = np.arange(n)
    tc = np.cos((2*k + 1)*np.pi/(2*n))            # nós de Chebyshev
    pc = np.polynomial.Polynomial.fit(tc, f(tc), deg=n-1).convert()
    cs = CubicSpline(tn, f(tn))                     # spline nos mesmos nós igualmente espaçados
    e_nos = np.max(np.abs(pn(tn) - f(tn)))
    e_eq  = np.max(np.abs(pn(t_plot) - f(t_plot)))
    e_ch  = np.max(np.abs(pc(t_plot) - f(t_plot)))
    e_sp  = np.max(np.abs(cs(t_plot) - f(t_plot)))
    print(f"{n:>3} | {e_nos:12.1e} | {e_eq:9.4f} | {e_ch:9.4f} | {e_sp:9.4f}")

n = 15
tn = np.linspace(-1, 1, n); k = np.arange(n); tc = np.cos((2*k + 1)*np.pi/(2*n))
plt.figure(figsize=(9, 5))
plt.plot(t_plot, f(t_plot), "k", lw=2, label="Função de referência")
plt.plot(t_plot, np.polynomial.Polynomial.fit(tn, f(tn), deg=n-1).convert()(t_plot), label="15 nós igualmente espaçados")
plt.plot(t_plot, np.polynomial.Polynomial.fit(tc, f(tc), deg=n-1).convert()(t_plot), "--", label="15 nós de Chebyshev")
plt.plot(t_plot, CubicSpline(tn, f(tn))(t_plot), ":", lw=2, label="Spline cúbica (15 nós)")
plt.ylim(-0.5, 1.2); plt.grid(True); plt.legend(); plt.title("Alternativas ao polinômio de grau alto")
plt.xlabel("x"); plt.ylabel("y"); plt.show()
```

## Seção 11
**Síntese do grupo:**

Interpolar é construir uma curva que passa exatamente pelos dados e usá-la para estimar valores **dentro** do intervalo dos nós; extrapolar é usar essa curva **fora** do intervalo, onde não há dados que sustentem o resultado. Com $m$ nós distintos existe um único polinômio de grau $\le m-1$: para os 4 pontos do primeiro exemplo obtivemos $P(x) = x^2 - 2x + 1$ (grau efetivo 2, com um resíduo de $10^{-16}$ no termo cúbico). NumPy, GeoGebra e WolframAlpha concordaram na expressão e nos valores $P(-1{,}5) = 6{,}25$ e $P(0{,}5) = 0{,}25$; as diferenças foram só de forma de apresentação (Newton × potências) e de arredondamento numérico. Ao alterar em $+0{,}5$ a medição em $x = 2$ do conjunto B, a estimativa em $x = 1{,}5$ passou de $2{,}3125$ para $2{,}59375$, e a curva mudou em todo o intervalo, não só perto do ponto alterado. Uma limitação importante é o fenômeno de Runge: com muitos nós igualmente espaçados, o polinômio de grau alto oscila nas extremidades e o erro aumenta com mais pontos. Usaríamos interpolação, por exemplo, para estimar a temperatura de um processo entre duas leituras próximas de um sensor confiável, com polinômio de grau baixo e sem extrapolar. Com muitos dados ou dados ruidosos, preferiríamos splines ou regressão.
