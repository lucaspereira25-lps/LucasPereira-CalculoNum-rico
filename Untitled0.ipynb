```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import pandas as pd

# ==============================================================================
# PARTE 1: IMPLEMENTAÇÃO DOS MÉTODOS
# ==============================================================================

def bissecao(f, a, b, tol=1e-6, max_iter=100):
    if f(a) * f(b) > 0:
        return None, 0, [], "Erro: f(a)*f(b) > 0 (Sem garantia de raiz no intervalo)"

    historico = []
    x_old = a

    for i in range(1, max_iter + 1):
        x = (a + b) / 2.0
        historico.append((a, b, x))

        if i > 1 and abs(x - x_old) < tol and abs(f(x)) < tol:
            return x, i, historico, "Convergiu"
        if abs(f(x)) < tol:
            return x, i, historico, "Convergiu"

        if f(a) * f(x) < 0:
            b = x
        else:
            a = x

        x_old = x

    return x, max_iter, historico, "Falha: Atingiu o número máximo de iterações"


def newton(f, df, x0, tol=1e-6, max_iter=100):
    historico = [x0]
    xk = x0

    for i in range(1, max_iter + 1):
        df_val = df(xk)
        if abs(df_val) < 1e-12:
            return xk, i-1, historico, "Erro: Derivada nula ou próxima de zero"

        x_next = xk - f(xk) / df_val
        historico.append(x_next)

        if abs(x_next - xk) < tol and abs(f(x_next)) < tol:
            return x_next, i, historico, "Convergiu"

        xk = x_next

    return xk, max_iter, historico, "Falha: Atingiu o número máximo de iterações"


def secante(f, x0, x1, tol=1e-6, max_iter=100):
    historico = [x0, x1]

    for i in range(1, max_iter + 1):
        f0, f1 = f(x0), f(x1)
        den = f1 - f0

        if abs(den) < 1e-12:
            return x1, i-1, historico, "Erro: Denominador nulo ou próximo de zero"

        x_next = x1 - f1 * (x1 - x0) / den
        historico.append(x_next)

        if abs(x_next - x1) < tol and abs(f(x_next)) < tol:
            return x_next, i, historico, "Convergiu"

        x0, x1 = x1, x_next

    return x1, max_iter, historico, "Falha: Atingiu o número máximo de iterações"


# ==============================================================================
# DEFINIÇÃO DAS FUNÇÕES E DERIVADAS
# ==============================================================================
def f1(x): return x**2 - 2
def df1(x): return 2*x

def f2(x): return x**3 - x - 2
def df2(x): return 3*x**2 - 1

def f3(x): return np.exp(-x) - x
def df3(x): return -np.exp(-x) - 1

def f4(x): return x**3 - 2*x + 2
def df4(x): return 3*x**2 - 2


# ==============================================================================
# PARTE 2: APLICAÇÃO, EXECUÇÃO E TABELA DE RESULTADOS
# ==============================================================================
configuracoes = [
    {"nome": "f1(x) = x^2 - 2", "f": f1, "df": df1, "biss": [1.0, 2.0], "newton": 1.5, "sec": (1.0, 2.0)},
    {"nome": "f2(x) = x^3 - x - 2", "f": f2, "df": df2, "biss": [1.0, 2.0], "newton": 1.5, "sec": (1.0, 2.0)},
    {"nome": "f3(x) = e^-x - x", "f": f3, "df": df3, "biss": [0.0, 1.0], "newton": 0.5, "sec": (0.0, 1.0)},
    {"nome": "f4(x) = x^3 - 2x + 2", "f": f4, "df": df4, "biss": [-2.0, -1.0], "newton": -2.0, "sec": (-2.0, -1.0)}
]

tabela_dados = []

for cfg in configuracoes:
    f, df, nome = cfg["f"], cfg["df"], cfg["nome"]

    # 1. Bisseção
    a, b = cfg["biss"]
    r, it, hist, st = bissecao(f, a, b)
    res = abs(f(r)) if r is not None else None
    tabela_dados.append([nome, "Bisseção", f"[{a}, {b}]", f"{r:.6f}" if r else "N/A", f"{res:.2e}" if res else "N/A", f"{it} / {st}"])

    # 2. Newton
    x0 = cfg["newton"]
    r, it, hist, st = newton(f, df, x0)
    res = abs(f(r)) if r is not None else None
    tabela_dados.append([nome, "Newton", f"x0 = {x0}", f"{r:.6f}" if r else "N/A", f"{res:.2e}" if res else "N/A", f"{it} / {st}"])

    # 3. Secante
    s0, s1 = cfg["sec"]
    r, it, hist, st = secante(f, s0, s1)
    res = abs(f(r)) if r is not None else None
    tabela_dados.append([nome, "Secante", f"({s0}, {s1})", f"{r:.6f}" if r else "N/A", f"{res:.2e}" if res else "N/A", f"{it} / {st}"])

df_resultados = pd.DataFrame(tabela_dados, columns=[
    "Função", "Método", "Valores iniciais", "Raiz aproximada", "|f(x)| final", "Iterações/resultado"
])

print("\n--- TABELA DE RESULTADOS (PARTE 2) ---")
print(df_resultados.to_string(index=False))


# ==============================================================================
# PARTE 3: ANIMAÇÃO GRÁFICA E GERAÇÃO DOS GIFS
# ==============================================================================

def criar_animacao_bissecao(f, a, b, filename="bissecao.gif"):
    r, it, hist, status = bissecao(f, a, b)
    fig, ax = plt.subplots(figsize=(8, 5))
    x_vals = np.linspace(a - 0.5, b + 0.5, 300)

    def update(frame):
        ax.clear()
        ca, cb, cx = hist[frame]
        ax.plot(x_vals, f(x_vals), 'b-', label='f(x)')
        ax.axhline(0, color='black', linewidth=1)
        ax.axvspan(ca, cb, color='yellow', alpha=0.3, label=f'Intervalo [{ca:.4f}, {cb:.4f}]')
        ax.plot(cx, f(cx), 'ro', label=f'x_{frame+1} = {cx:.5f}')
        ax.set_title(f'Bisseção - Iteração {frame+1}/{len(hist)} | |f(x)| = {abs(f(cx)):.2e}')
        ax.set_xlabel('x')
        ax.set_ylabel('f(x)')
        ax.legend(loc='upper left')
        ax.grid(True)

    anim = animation.FuncAnimation(fig, update, frames=len(hist), interval=1000)
    anim.save(filename, writer='pillow')
    plt.close()

def criar_animacao_newton(f, df, x0, filename="newton.gif"):
    r, it, hist, status = newton(f, df, x0)
    fig, ax = plt.subplots(figsize=(8, 5))
    x_vals = np.linspace(min(hist) - 0.5, max(hist) + 0.5, 300)

    def update(frame):
        ax.clear()
        xk = hist[frame]
        ax.plot(x_vals, f(x_vals), 'b-', label='f(x)')
        ax.axhline(0, color='black', linewidth=1)
        ax.plot(xk, f(xk), 'ro', label=f'x_{frame} = {xk:.5f}')

        if frame < len(hist) - 1:
            tangente = f(xk) + df(xk) * (x_vals - xk)
            ax.plot(x_vals, tangente, 'r--', label='Reta Tangente')
            ax.plot(hist[frame+1], 0, 'go', label=f'Próximo x = {hist[frame+1]:.5f}')

        ax.set_title(f'Newton - Iteração {frame} | |f(x)| = {abs(f(xk)):.2e}')
        ax.set_xlabel('x')
        ax.set_ylabel('f(x)')
        ax.legend(loc='upper left')
        ax.grid(True)

    anim = animation.FuncAnimation(fig, update, frames=len(hist), interval=1000)
    anim.save(filename, writer='pillow')
    plt.close()

def criar_animacao_secante(f, x0, x1, filename="secante.gif"):
    r, it, hist, status = secante(f, x0, x1)
    fig, ax = plt.subplots(figsize=(8, 5))
    x_vals = np.linspace(min(hist) - 0.5, max(hist) + 0.5, 300)

    def update(frame):
        ax.clear()
        ax.plot(x_vals, f(x_vals), 'b-', label='f(x)')
        ax.axhline(0, color='black', linewidth=1)

        x_a = hist[frame]
        x_b = hist[frame+1]

        ax.plot([x_a, x_b], [f(x_a), f(x_b)], 'ro', label=f'Pontos x_{frame}, x_{frame+1}')
        m = (f(x_b) - f(x_a)) / (x_b - x_a)
        sec_line = f(x_b) + m * (x_vals - x_b)
        ax.plot(x_vals, sec_line, 'g--', label='Reta Secante')

        if frame + 2 < len(hist):
            ax.plot(hist[frame+2], 0, 'mo', label=f'Novo x = {hist[frame+2]:.5f}')

        ax.set_title(f'Secante - Passo {frame+1} | |f(x_b)| = {abs(f(x_b)):.2e}')
        ax.set_xlabel('x')
        ax.set_ylabel('f(x)')
        ax.legend(loc='upper left')
        ax.grid(True)

    anim = animation.FuncAnimation(fig, update, frames=len(hist)-1, interval=1000)
    anim.save(filename, writer='pillow')
    plt.close()

print("\nA gerar arquivos GIF das animações...")
criar_animacao_bissecao(f1, 1.0, 2.0, "animacao_bissecao.gif")
criar_animacao_newton(f2, df2, 1.5, "animacao_newton.gif")
criar_animacao_secante(f3, 0.0, 1.0, "animacao_secante.gif")
print("Animações salvas com sucesso!")
```
