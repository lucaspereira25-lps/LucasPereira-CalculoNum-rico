# Tópico 5

Resolução do Tópico 5 (Parte C – Vazão em um Tubo de Pequeno Diâmetro)Dados de Referência:$r = 0{,}80\text{ mm} = 8{,}0 \times 10^{-4}\text{ m}$$L = 0{,}20\text{ m}$$\mu = 3{,}5 \times 10^{-3}\text{ Pa}\cdot\text{s}$$\Delta p = 1200\text{ Pa}$Fórmula de Hagen–Poiseuille:$$Q = \frac{\pi r^4 \Delta p}{8 \mu L}$$Cálculo da vazão de referência:$$Q_{\text{ref}} = \frac{\pi \cdot (8{,}0 \times 10^{-4})^4 \cdot 1200}{8 \cdot (3{,}5 \times 10^{-3}) \cdot 0{,}20} \approx 2{,}7646015 \times 10^{-7} \text{ m}^3/\text{s}$$C.1 Precisão das VariáveisComo os valores originais fornecidos ($r = 0{,}80\text{ mm}$, $\mu = 0{,}0035\text{ Pa}\cdot\text{s}$, $\Delta p = 1200\text{ Pa}$) já coincidem exatamente com suas respectivas aproximações pedidas (1 casa decimal para $r$, 3 casas para $\mu$ e arredondamento para centenas em $\Delta p$), os cenários de 1 a 5 resultam em erro zero.




image.png


```python
import matplotlib.pyplot as plt
import numpy as np

# Raio variando de 0.70 a 0.90 mm (passo de 0.01 mm)
r_mm = np.linspace(0.70, 0.90, 21)
r_m = r_mm * 1e-3

L = 0.20
mu = 3.5e-3
dp = 1200.0

# Q Referência
Q_ref = (np.pi * (r_m**4) * dp) / (8 * mu * L)

# Aproximações de r (1 casa decimal em mm)
r_arr_mm = np.round(r_mm, 1)
r_trunc_mm = np.trunc(r_mm * 10) / 10

Q_arr = (np.pi * ((r_arr_mm * 1e-3)**4) * dp) / (8 * mu * L)
Q_trunc = (np.pi * ((r_trunc_mm * 1e-3)**4) * dp) / (8 * mu * L)

# Erros para Arredondamento
ea_arr = np.abs(Q_ref - Q_arr)
er_arr = ea_arr / Q_ref
ep_arr = er_arr * 100

# Erros para Truncamento
ea_trunc = np.abs(Q_ref - Q_trunc)
er_trunc = ea_trunc / Q_ref
ep_trunc = er_trunc * 100

plt.figure(figsize=(12, 10))

# 1. Q em função de r
plt.subplot(2, 2, 1)
plt.plot(r_mm, Q_ref*1e7, 'k-', label='Q Referência')
plt.plot(r_mm, Q_arr*1e7, 'b--', label='Q Arredondado')
plt.plot(r_mm, Q_trunc*1e7, 'r:', label='Q Truncado')
plt.xlabel('Raio r (mm)')
plt.ylabel('Vazão Q ($10^{-7} m^3/s$)')
plt.title('1. Vazão Q em função de r')
plt.legend()
plt.grid(True)

# 2. Erro Absoluto
plt.subplot(2, 2, 2)
plt.plot(r_mm, ea_arr*1e7, 'b-o', label='Arredondamento')
plt.plot(r_mm, ea_trunc*1e7, 'r-s', label='Truncamento')
plt.xlabel('Raio r (mm)')
plt.ylabel('Erro Absoluto ($10^{-7} m^3/s$)')
plt.title('2. Erro Absoluto de Q')
plt.legend()
plt.grid(True)

# 3. Erro Relativo
plt.subplot(2, 2, 3)
plt.plot(r_mm, er_arr, 'b-o', label='Arredondamento')
plt.plot(r_mm, er_trunc, 'r-s', label='Truncamento')
plt.xlabel('Raio r (mm)')
plt.ylabel('Erro Relativo ($E_r$)')
plt.title('3. Erro Relativo de Q')
plt.legend()
plt.grid(True)

# 4. Erro Percentual
plt.subplot(2, 2, 4)
plt.plot(r_mm, ep_arr, 'b-o', label='Arredondamento')
plt.plot(r_mm, ep_trunc, 'r-s', label='Truncamento')
plt.xlabel('Raio r (mm)')
plt.ylabel('Erro Percentual ($E_{\\%}$)')
plt.title('4. Erro Percentual de Q')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

Explicação da Amplificação do Erro no RaioPela equação de Hagen–Poiseuille, a vazão $Q$ é diretamente proporcional à quarta potência do raio ($Q \propto r^4$).Utilizando a aproximação por diferencial para estimar a variação relativa de $Q$ em relação a $r$:$$\frac{dQ}{dr} = \frac{4 \pi r^3 \Delta p}{8 \mu L} = \frac{4}{r} \left(\frac{\pi r^4 \Delta p}{8 \mu L}\right) = \frac{4 Q}{r}$$Dividindo ambos os lados por $Q$, obtém-se a relação entre as variações relativas:$$\frac{dQ}{Q} = 4 \frac{dr}{r} \implies E_{r, Q} \approx 4 \cdot E_{r, r}$$O expoente quatro atua como um fator multiplicador direto da variação relativa. Isso significa que qualquer pequena imprecisão ou erro percentual cometida na medição do raio $r$ será amplificada aproximadamente quatro vezes no valor final da vazão $Q$.

Resolução do Tópico 6 (Parte D – Escoamento em um Nanocateter)Valores de Referência do Nanocateter:

*   $r = 50\text{ nm} = 50 \times 10^{-9}\text{ m}$
*   $L = 10\ \mu\text{m} = 10 \times 10^{-6}\text{ m}$
*  $\mu = 3{,}5 \times 10^{-3}\text{ Pa}\cdot\text{s}$
*   $\Delta p = 5000\text{ Pa}$

Vazão nominal de referência:$$Q(r) = \frac{\pi (50 \times 10^{-9})^4 \cdot 5000}{8 (3{,}5 \times 10^{-3}) (10 \times 10^{-6})} \approx 3{,}51629 \times 10^{-16} \text{ m}^3/\text{s}$$

D.1 Erro Geométrico & Código Python

```python
import matplotlib.pyplot as plt
import numpy as np

# Dados nominais
r = 50e-9
L = 10e-6
mu = 3.5e-3
dp = 5000.0

def calc_Q(r_val):
    return (np.pi * (r_val**4) * dp) / (8 * mu * L)

Q_ref = calc_Q(r)

# Variação de delta_r de 0.1 a 5.0 nm
dr_nm = np.arange(0.1, 5.1, 0.1)
dr_m = dr_nm * 1e-9

r_minus = r - dr_m
r_plus = r + dr_m

Q_minus = calc_Q(r_minus)
Q_plus = calc_Q(r_plus)

# Erro máximo (ocorre no limite superior r_plus)
E_a_max = np.abs(Q_plus - Q_ref)
E_r_max = E_a_max / Q_ref
E_p_max = E_r_max * 100

# Aproximação linear 4 * (dr / r)
aprox_linear = 4 * (dr_m / r)

# Plot dos gráficos
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.plot(dr_nm, Q_minus*1e16, label='$Q(r_-)$')
plt.plot(dr_nm, np.full_like(dr_nm, Q_ref*1e16), 'k--', label='$Q(r)$ nominal')
plt.plot(dr_nm, Q_plus*1e16, label='$Q(r_+)$')
plt.xlabel('$\Delta r$ (nm)')
plt.ylabel('Vazão $Q$ ($10^{-16}\ m^3/s$)')
plt.title('1. Vazões $Q(r_-)$, $Q(r)$ e $Q(r_+)$')
plt.legend()
plt.grid(True)

plt.subplot(2, 2, 2)
plt.plot(dr_nm, E_a_max*1e16, 'r-')
plt.xlabel('$\Delta r$ (nm)')
plt.ylabel('Erro Absoluto Máx ($10^{-16}\ m^3/s$)')
plt.title('2. Erro Absoluto Máximo de $Q$')
plt.grid(True)

plt.subplot(2, 2, 3)
plt.plot(dr_nm, E_r_max, 'b-', label='Erro Relativo Exato')
plt.plot(dr_nm, aprox_linear, 'k--', label='Aproximação Linear ($4 \\Delta r / r$)')
plt.xlabel('$\Delta r$ (nm)')
plt.ylabel('Erro Relativo ($E_r$)')
plt.title('3. Erro Relativo Máximo vs Aprox. Linear')
plt.legend()
plt.grid(True)

plt.subplot(2, 2, 4)
plt.plot(dr_nm, E_p_max, 'g-')
plt.xlabel('$\Delta r$ (nm)')
plt.ylabel('Erro Percentual Máximo ($E_{\\%}$)')
plt.title('4. Erro Percentual Máximo de $Q$')
plt.grid(True)

plt.tight_layout()
plt.show()
```

Comparação com a Aproximação Linear:A aproximação linear $E_r \approx 4 \frac{\Delta r}{r}$ acompanha muito bem a tendência para pequenos erros (por exemplo, em $\Delta r = 0{,}1\text{ nm}$, o erro exato é de $0{,}806\%$ e a aproximação linear prevê $0{,}800\%$). Contudo, à medida que $\Delta r$ cresce até $5{,}0\text{ nm}$ ($10\%$ do raio), os termos de ordem superior da expansão por Binômio de Newton de $(1 + \Delta r/r)^4 - 1$ passam a ter peso, fazendo com que o erro real ($46{,}41\%$) supere a estimativa linear ($40{,}00\%$).

Análise de Perda de Significância por Cancelamento ($\delta_1$ vs $\delta_2$):
Ao calcular variações extremamente pequenas (por exemplo, $\Delta r/r \le 10^{-7}$):
*   A fórmula $\delta_1 = \frac{Q(r+\Delta r) - Q(r)}{Q(r)}$ sofre de cancelamento catastrófico, pois subtrai dois números praticamente idênticos ($Q(r+\Delta r) \approx Q(r)$), resultando no descarte de algarismos significativos e gerando ruído de arredondamento ou erro nulo.
*   A fórmula $\delta_2 = \left(1 + \frac{\Delta r}{r}\right)^4 - 1$ é algebricamente equivalente, mas quando $\Delta r/r$ é muito pequeno, a soma $1 + \Delta r/r$ em float32 perde a parte decimal pela precisão da máquina. No entanto, sua expansão analítica direta evita a subtração direta de grandes números computados de vazão, mantendo maior estabilidade numérica.

D.3 Fenômeno FísicoÀ medida que a escala geométrica diminui até a dimensão nanométrica, a razão entre a área de superfície e o volume ($A/V = 2/r$) cresce inversamente proporcional ao raio, tornando os efeitos viscosos e a atrito de parede vertiginosamente dominantes sobre os efeitos inerciais. Em um nanocateter, o alto atrito junto às paredes e os efeitos de camada limite ocupam uma fração expressiva do volume total do fluido, exigindo que variações de frações de nanômetro na geometria alterem drasticamente o perfil de velocidade e a vazão. Isso demonstra que um cálculo pode apresentar alta precisão numérica (muitas casas decimais), mas ser fisicamente inadequado se a aproximação do meio contínuo da equação de Hagen–Poiseuille ignorar fenômenos de escala como o deslizamento fluido-parede e a estruturação molecular do solvente.



# Tópico 8
O código em Python a seguir utiliza as bibliotecas matplotlib e Pillow (ou a extensão de animação integrada do Matplotlib) para gerar um GIF animado com 50 quadros. O raio varia continuamente de $45\text{ nm}$ a $55\text{ nm}$.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter

# 1. Parâmetros físicos
L = 10e-6         # 10 um
mu = 3.5e-3       # 3.5 mPa.s
dp = 5000.0       # 5000 Pa
r_ref = 50.0      # 50 nm

def calcular_vazao(r_nm):
    r_m = r_nm * 1e-9
    return (np.pi * (r_m**4) * dp) / (8 * mu * L)

Q_ref = calcular_vazao(r_ref)

# 2. Dados da animação
num_frames = 50
r_vals = np.linspace(45, 55, num_frames)
Q_vals = calcular_vazao(r_vals)

# 3. Criação do Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

def update(frame):
    ax1.cla()
    ax2.cla()

    r_atual = r_vals[frame]
    Q_atual = Q_vals[frame]
    erro_perc = ((Q_atual - Q_ref) / Q_ref) * 100

    # Painel 1: Seção Transversal
    circulo_ref = plt.Circle((0, 0), r_ref, color='gray', fill=False, ls='--', label='Ref. (50 nm)')
    circulo_atual = plt.Circle((0, 0), r_atual, color='tab:blue', alpha=0.5, label=f'Atual ({r_atual:.1f} nm)')

    ax1.add_patch(circulo_ref)
    ax1.add_patch(circulo_atual)
    ax1.set_xlim(-60, 60)
    ax1.set_ylim(-60, 60)
    ax1.set_aspect('equal')
    ax1.set_title(f'Seção Transversal (r = {r_atual:.1f} nm)')
    ax1.grid(True, ls=':')
    ax1.legend(loc='upper right')

    # Painel 2: Curva Q x r
    ax2.plot(r_vals[:frame+1], Q_vals[:frame+1] * 1e16, 'b-')
    ax2.plot(r_atual, Q_atual * 1e16, 'ro', label='Atual')
    ax2.axvline(x=50, color='gray', ls='--')

    ax2.set_xlim(44.5, 55.5)
    ax2.set_ylim(min(Q_vals)*1e16 * 0.9, max(Q_vals)*1e16 * 1.1)
    ax2.set_xlabel('Raio r (nm)')
    ax2.set_ylabel('Vazão Q ($10^{-16} m^3/s$)')
    ax2.set_title(f'Vazão: {Q_atual*1e16:.2f} | Erro: {erro_perc:+.1f}%')
    ax2.grid(True)

ani = FuncAnimation(fig, update, frames=num_frames, interval=100)

# Salva o arquivo localmente
ani.save('nanocateter.gif', writer=PillowWriter(fps=10))
plt.close(fig)

print("GIF gerado com sucesso: 'nanocateter.gif'")
```
